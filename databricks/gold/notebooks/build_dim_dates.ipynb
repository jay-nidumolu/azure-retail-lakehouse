```python
from pyspark.sql import functions as F
```

```python
storage_account= "storageretaillakehouse"

spark.conf.set(
    f"fs.azure.account.key.{storage_account}.dfs.core.windows.net",
    "<YOUR KEY>"
)
```

```python
gold_path = f"abfss://gold@{storage_account}.dfs.core.windows.net/dimensions/dim_dates"
```

```python

dates = spark.sql("""
SELECT  explode(
    sequence(
        to_date('2016-01-01'),
        to_date('2020-12-31'),
        interval 1 day
    )
    ) AS date
""")

```

```python
dim_dates = (
    dates
    .withColumn("date_key",
                F.date_format("date","yyyyMMdd"))
    .withColumn("year",
                F.year("date"))
    .withColumn("quarter",
                F.quarter("date"))
    .withColumn("month",
                F.month("date"))
    .withColumn("month_name",
                F.date_format("date","MMMM"))
    .withColumn("week",
                F.weekofyear("date"))
    .withColumn("day",
                F.dayofmonth("date"))
)


```

```python


dim_dates.write.mode("overwrite").save(gold_path)
```
